```python
#program to write the keys as numbers between 1 to 15 amd the values as their square
my_dict = {}
for i in range(1,16):
    my_dict[i] == i**2
print(my_dict)
```

```python
#program to print the total of all the values
name = {
    101: "Ankit",
    102: "Mohit",
    103: "Rohit"
}
total = 0
for number in name.keys():
    total = total + number
print(total)
```
