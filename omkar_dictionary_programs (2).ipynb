```python
#for printing the decinding order of the upper program
my_dict = {
    "Amit":85,
    "Rohit":39,
    "Priya":90,
}
sorted_dict = dict(sorted(my_dict.items() , key=lambda item:item[1]))
sorted_dict.sort(reverse=True)
print("Dictionary sorted by values")
print(sorted_dict)
```

```python
#python prgram to count the frequency of a item in the dictionary
numbers = [1,2,3,4,2,1,2,2]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] =  frequency[num] + 1
    else:
        frequency[num] = 1

print("The frequency of the numbers is " , frequency)
```
